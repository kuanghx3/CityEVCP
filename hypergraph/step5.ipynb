```python
import pandas as pd
import numpy as np
```

```python
poi_df = pd.read_csv("SZInformation_POItype.csv",header=0,index_col=0)
poi_df.head()
```

```python
hg_list = []
for i in range (10):
    a = np.zeros((247))
    for j in range(247):
        if poi_df["POI_type"][j] == i:
            NO = poi_df["NO"][j]
            a[NO] = 1
    hg_list.append(a)
    
```

```python
hg = np.array(hg_list)
hg = hg.T
print(hg.shape)
pd.DataFrame(data=hg).to_csv("SZ247_hypergraph_2.csv")
```

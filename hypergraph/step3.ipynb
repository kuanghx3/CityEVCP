```python
import pandas as pd
import numpy as np
import math
```

```python
df = pd.read_csv("SZ_area_POI.csv",header=0,index_col=0)
data_np = np.array(df)
data_np_new = np.delete(data_np,0,1)
```

```python
print(data_np_new.shape)
len = data_np_new.shape[0]
```

```python
TF = np.zeros_like(data_np_new)
for i in range(len):
    count = np.sum(data_np_new[i])
    for j in range (data_np_new.shape[1]):
        TF[i,j] = data_np_new[i,j]/count
```

```python
IDF = np.zeros_like(data_np_new)
nozerocount = np.count_nonzero(data_np_new,axis=1)
for i in range(data_np_new.shape[1]):
    for j in range(len):
        IDF[j,i] = math.log(len/nozerocount[i])
TF_IDF = TF*IDF    
```

```python
print(TF_IDF.shape)
```

```python
from sklearn.cluster import KMeans

y_pred = KMeans(n_clusters=6, random_state=9).fit_predict(TF_IDF)
print(y_pred.shape)
```

```python
info = pd.read_csv("Shenzhen_Pile_Information.csv",header=0)
y_pred_6 = KMeans(n_clusters=6, random_state=9,n_init= 'auto').fit_predict(TF_IDF)
info["n_6"] = y_pred_6
y_pred_8 = KMeans(n_clusters=8, random_state=9,n_init= 'auto').fit_predict(TF_IDF)
info["n_8"] = y_pred_8
y_pred_10 = KMeans(n_clusters=10, random_state=9,n_init= 'auto').fit_predict(TF_IDF)
info["n_10"] = y_pred_10
y_pred_12 = KMeans(n_clusters=12, random_state=9,n_init= 'auto').fit_predict(TF_IDF)
info["n_12"] = y_pred_12
y_pred_14 = KMeans(n_clusters=14, random_state=9,n_init= 'auto').fit_predict(TF_IDF)
info["n_14"] = y_pred_14
```

```python
info.to_csv("SZEV_info_cluster.csv")
```
